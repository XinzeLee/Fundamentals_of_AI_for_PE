## Field temperature regression with a residual feedforward network

This notebook follows the **good-practice** workflow from `Good_Practices/good_practice_NN.ipynb`, applied to **3D thermal field** data stored as CSV files in this folder.

**Task:** predict temperature **T** at spatial samples \((x,y,z)\) given **dissipated power** (`loss`, decoded from the filename, in W) and **ambient temperature** `Tamb` (from the filename, in °C). Columns `x`, `y`, `z`, and `T` come from each CSV.

**Practices emphasized (same spirit as the house-price tutorial):**

- Ensure the model has enough capacity to **overfit** the training split when debugging; then regularize or simplify if needed.
- Use **normalization**: `StandardScaler` on inputs (and on **T** for stable MSE), fit **only on training** data.
- Split data into **train / validation / test**; tune using validation, report final numbers on test.
- Train with **mini-batches** and inspect **learning curves**.
- Use a **learning-rate scheduler** (`ReduceLROnPlateau`) when validation loss stalls.
- Keep the **best model weights in memory** (lowest validation loss) and **reload** that state for evaluation—no local checkpoint file is written.

**Model:** a **feedforward network (FNN)** with **residual (skip) blocks** so deeper stacks train more reliably for smooth field-type regression.

### 1. Resolve paths, discover CSV files, parse operating conditions

Filenames follow patterns like `Tfield_loss0p5W_Tamb20C_downsampled.csv`:

- **`loss`:** token after `loss` and before `W`; `0p5` is interpreted as **0.5** watts (decimal encoded with `p`).
- **`Tamb`:** integer °C after `Tamb` and before `C`.

If a file does not match this pattern, it is skipped with a warning.

**Working directory:** In Jupyter, the kernel’s current directory is typically the folder that contains this notebook (`Field_Data`). All `*.csv` files in that same folder are loaded (sorted by name). The Colab setup cell above also `%cd`s into this folder so behavior matches.


```python
import re
import warnings
from pathlib import Path

import numpy as np
import pandas as pd

# Look inside the "cap_Tfield" folder within the current directory.
DATA_DIR = (Path.cwd() / "cap_Tfield").resolve()
print("DATA_DIR (loads sorted *.csv here):", DATA_DIR)


def parse_loss_tamb_from_filename(path: Path) -> tuple[float, float]:
    """Parse dissipated power (W) and ambient Tamb (°C) from e.g. ..._loss0p5W_Tamb20C_...."""
    stem = path.stem
    m_loss = re.search(r"loss(?P<loss>[\d.p]+)W", stem, re.IGNORECASE)
    m_tamb = re.search(r"Tamb(?P<tamb>\d+)C", stem, re.IGNORECASE)
    if not m_loss or not m_tamb:
        raise ValueError(f"Missing loss*/W or Tamb*C in filename: {path.name}")
    loss_token = m_loss.group("loss").replace("p", ".")
    loss_w = float(loss_token)
    tamb_c = float(m_tamb.group("tamb"))
    return loss_w, tamb_c


def load_all_field_csvs(data_dir: Path) -> pd.DataFrame:
    rows = []
    for csv_path in sorted(data_dir.glob("*.csv")):
        try:
            loss_w, tamb_c = parse_loss_tamb_from_filename(csv_path)
        except ValueError as e:
            warnings.warn(f"Skip {csv_path.name}: {e}")
            continue
        df = pd.read_csv(csv_path)
        need = {"x", "y", "z", "T"}
        if not need.issubset(df.columns):
            warnings.warn(f"Skip {csv_path.name}: need columns {need}")
            continue
        df = df[["x", "y", "z", "T"]].copy()
        df["loss_w"] = loss_w
        df["tamb_c"] = tamb_c
        df["source_file"] = csv_path.name
        rows.append(df)
    if not rows:
        raise RuntimeError(f"No valid CSVs in {data_dir}")
    return pd.concat(rows, ignore_index=True)


full_df = load_all_field_csvs(DATA_DIR)
print(full_df.head())
print("Shapes:", full_df.shape)
print(full_df["source_file"].value_counts())
```

### 2. Train / validation / test split (by CSV file) and scaling

Splits are defined **per CSV file** (not by random rows), so whole operating cases stay in one split—ready for additional datasets as you add files.

- **Three or more files:** ~**60% / 20% / 20%** of files to train / validation / test (`train_test_split`, `random_state=42`).
- **Two files:** one file is **train**, the other **test**; **validation** is a **25% row holdout** from the training file only.
- **Single file:** **60% / 20% / 20%** **row** split inside that file (degenerate case when only one CSV exists).

`StandardScaler` is **fit on training only**; validation and test use the training statistics. Targets **T** are scaled for MSE; plots convert back to °C.


```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

feature_cols = ["x", "y", "z", "loss_w", "tamb_c"]
X = full_df[feature_cols].to_numpy(dtype=np.float64)
y = full_df["T"].to_numpy(dtype=np.float64).reshape(-1, 1)

# Shuffle file names for split reproducibility
file_names = full_df["source_file"].unique()
rng = np.random.RandomState(42)
shuffled_files = file_names.copy()
rng.shuffle(shuffled_files)
file_names = shuffled_files
n_files = len(file_names)

if n_files >= 3:
    # Assign files to train (60%), validation (20%), and test (20%) splits
    train_names, val_names, test_names = np.split(
        file_names,
        [
            int(0.6 * n_files),
            int(0.8 * n_files)
        ]
    )
    train_mask = full_df["source_file"].isin(train_names).to_numpy()
    val_mask = full_df["source_file"].isin(val_names).to_numpy()
    test_mask = full_df["source_file"].isin(test_names).to_numpy()
elif n_files == 2:
    # One file for train, one for test; validation is 25% row holdout from train file
    train_names, test_names = file_names[:1], file_names[1:]
    val_names = []
    pool_train = full_df["source_file"].isin(train_names).to_numpy()
    idx_train_rows = np.where(pool_train)[0]
    sub_tr, sub_va = train_test_split(idx_train_rows, test_size=0.25, random_state=42)
    train_mask = np.zeros(len(full_df), dtype=bool)
    train_mask[sub_tr] = True
    val_mask = np.zeros(len(full_df), dtype=bool)
    val_mask[sub_va] = True
    test_mask = full_df["source_file"].isin(test_names).to_numpy()
else:
    # Single file case: row split 60/20/20
    idx = np.arange(len(full_df))
    tr, va_te = train_test_split(idx, test_size=0.4, random_state=42)
    va, te = train_test_split(va_te, test_size=0.5, random_state=42)
    train_mask = np.zeros(len(full_df), dtype=bool)
    val_mask = np.zeros(len(full_df), dtype=bool)
    test_mask = np.zeros(len(full_df), dtype=bool)
    train_mask[tr] = True
    val_mask[va] = True
    test_mask[te] = True
    train_names = val_names = test_names = file_names

X_train, y_train = X[train_mask], y[train_mask]
X_val, y_val = X[val_mask], y[val_mask]
X_test, y_test = X[test_mask], y[test_mask]

print(f"CSV files ({n_files}):", file_names)
if n_files >= 3:
    print("Train files:", sorted(list(train_names)))
    print("Val files:", sorted(list(val_names)))
    print("Test files:", sorted(list(test_names)))
elif n_files == 2:
    print("Train file(s):", train_names, "| Test file(s):", test_names)
    print("(Validation = 25% rows from training file.)")
else:
    print("Single file: row-based 60/20/20 split.")

input_scaler = StandardScaler()
X_train_n = input_scaler.fit_transform(X_train)
X_val_n = input_scaler.transform(X_val)
X_test_n = input_scaler.transform(X_test)

target_scaler = StandardScaler()
y_train_n = target_scaler.fit_transform(y_train)
y_val_n = target_scaler.transform(y_val)
y_test_n = target_scaler.transform(y_test)

print("Train:", X_train_n.shape, "Val:", X_val_n.shape, "Test:", X_test_n.shape)
```

### 3. PyTorch `Dataset`, residual FNN, and why residuals help

Each **residual block** implements $h \mapsto \sigma(\mathcal{F}(h) + h)$ with two linear layers and batch normalization, where $\mathcal{F}$ learns a **residual** correction. Identity shortcuts make it easier to train deeper stacks because layers can learn near-zero updates when depth is not needed—useful for smooth **field** regression.

A **stem** maps the 5D input $(x, y, z, \text{loss}, \text{Tamb})$ into the hidden width; several identical-width residual blocks follow; a final **linear head** outputs a scalar in **normalized T** space.


```python
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, TensorDataset

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)

torch.manual_seed(42)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(42)


class ResidualBlock(nn.Module):
    """Pre-norm style MLP sub-block with same input/output dim and identity skip."""

    def __init__(self, dim: int):
        super().__init__()
        self.fc1 = nn.Linear(dim, dim)
        self.bn1 = nn.BatchNorm1d(dim)
        self.fc2 = nn.Linear(dim, dim)
        self.bn2 = nn.BatchNorm1d(dim)
        self.act = nn.ReLU(inplace=True)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        identity = x
        out = self.fc1(x)
        out = self.bn1(out)
        out = self.act(out)
        out = self.fc2(out)
        out = self.bn2(out)
        return self.act(out + identity)


class ResidualFieldFNN(nn.Module):
    def __init__(self, in_dim: int = 5, hidden: int = 128, n_blocks: int = 4):
        super().__init__()
        self.stem = nn.Sequential(
            nn.Linear(in_dim, hidden),
            nn.BatchNorm1d(hidden),
            nn.ReLU(inplace=True),
        )
        self.blocks = nn.Sequential(*[ResidualBlock(hidden) for _ in range(n_blocks)])
        self.head = nn.Linear(hidden, 1)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        x = self.stem(x)
        x = self.blocks(x)
        return self.head(x)


def make_loaders(
    batch_size: int,
) -> tuple[DataLoader, DataLoader, np.ndarray, np.ndarray]:
    train_ds = TensorDataset(
        torch.from_numpy(X_train_n.astype(np.float32)),
        torch.from_numpy(y_train_n.astype(np.float32)),
    )
    val_ds = TensorDataset(
        torch.from_numpy(X_val_n.astype(np.float32)),
        torch.from_numpy(y_val_n.astype(np.float32)),
    )
    train_loader = DataLoader(train_ds, batch_size=batch_size, shuffle=True, drop_last=False)
    val_loader = DataLoader(val_ds, batch_size=batch_size, shuffle=False, drop_last=False)
    return train_loader, val_loader, X_test_n, y_test_n


BATCH_SIZE = 512
train_loader, val_loader, X_test_n, y_test_n = make_loaders(BATCH_SIZE)
if len(val_loader.dataset) == 0:
    raise ValueError("Validation set is empty — check CSV split logic.")
```

### 4. Training: MSE, Adam, `ReduceLROnPlateau`, best weights in memory

We minimize **MSE** on **scaled** T. **Huber** loss is a robust alternative if outliers appear; here the field samples are smooth, so MSE is appropriate.

After each epoch we:

1. Step **`ReduceLROnPlateau`** using validation loss.
2. If validation improves, store a **copy** of `model.state_dict()` (no file written).

At the end we **restore** the best state dict so plots and metrics use the early-stopped-best model, not the last epoch.


```python
import copy

import torch.optim as optim

EPOCHS = 400
LR = 1e-3

model = ResidualFieldFNN(in_dim=5, hidden=12, n_blocks=2).to(device)
criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=LR)
scheduler = optim.lr_scheduler.ReduceLROnPlateau(
    optimizer, mode="min", factor=0.5, patience=15
)

train_losses: list[float] = []
val_losses: list[float] = []
best_val = float("inf")
best_epoch = -1
best_model_state = None


@torch.no_grad()
def evaluate_loader(m: nn.Module, loader: DataLoader) -> float:
    m.eval()
    total_sq = 0.0
    total_n = 0
    for xb, yb in loader:
        xb = xb.to(device)
        yb = yb.to(device)
        pred = m(xb)
        total_sq += (pred - yb).pow(2).sum().item()
        total_n += yb.numel()
    m.train()
    return total_sq / max(total_n, 1)


for epoch in range(EPOCHS):
    model.train()
    epoch_train_sq = 0.0
    epoch_train_n = 0
    for xb, yb in train_loader:
        xb = xb.to(device)
        yb = yb.to(device)
        optimizer.zero_grad(set_to_none=True)
        pred = model(xb)
        loss = criterion(pred, yb)
        loss.backward()
        optimizer.step()
        epoch_train_sq += (pred - yb).pow(2).sum().item()
        epoch_train_n += yb.numel()

    train_mse = epoch_train_sq / max(epoch_train_n, 1)
    val_mse = evaluate_loader(model, val_loader)
    train_losses.append(train_mse)
    val_losses.append(val_mse)
    scheduler.step(val_mse)

    if val_mse < best_val:
        best_val = val_mse
        best_epoch = epoch
        best_model_state = copy.deepcopy(model.state_dict())

    if (epoch + 1) % 20 == 0 or epoch == 0:
        lr = optimizer.param_groups[0]["lr"]
        print(
            f"Epoch {epoch + 1}/{EPOCHS}  train_mse(norm)={train_mse:.6f}  val_mse(norm)={val_mse:.6f}  lr={lr:.2e}"
        )

print(f"Best val MSE (normalized T): {best_val:.6f} at epoch {best_epoch + 1}")

if best_model_state is not None:
    model.load_state_dict(best_model_state)
    print(f"Restored best model weights from epoch {best_epoch + 1} (in memory only).")
```

### 5. Accuracy plots and metrics (test set, °C)

We report **MAE**, **RMSE**, and **R²** in **physical units** after inverse-transforming predictions and targets with `target_scaler`. Plots:

- **Learning curves** (train vs. validation MSE in normalized space).
- **Predicted vs. actual** scatter with the $y = x$ line.
- **Residual histogram** of $T_{\mathrm{true}} - T_{\mathrm{pred}}$ (same definition as the 3-D figure).
- **3-D scatter** of test points $(x, y, z)$ with color showing $T_{\mathrm{true}} - T_{\mathrm{pred}}$.
- **Hotspot report:** the test sample with **largest true $T$** and its predicted temperature.
- **Optional:** 2-D scatter colored by `Tamb` for multi-condition datasets.


```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D  # noqa: F401
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

fig, ax = plt.subplots(1, 1, figsize=(7, 4))
ax.plot(train_losses, label="Train MSE (norm. T)")
ax.plot(val_losses, label="Val MSE (norm. T)")
ax.axvline(best_epoch, color="k", linestyle="--", alpha=0.5, label=f"Best epoch {best_epoch + 1}")
ax.set_xlabel("Epoch")
ax.set_ylabel("MSE")
ax.set_title("Learning curves (scaled target space)")
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()


@torch.no_grad()
def predict_numpy(m: nn.Module, Xn: np.ndarray) -> np.ndarray:
    m.eval()
    xb = torch.from_numpy(Xn.astype(np.float32)).to(device)
    out = m(xb).cpu().numpy()
    return out


y_test_pred_n = predict_numpy(model, X_test_n)
y_test_true = target_scaler.inverse_transform(y_test_n).ravel()
y_test_pred = target_scaler.inverse_transform(y_test_pred_n).ravel()

mae = mean_absolute_error(y_test_true, y_test_pred)
rmse = float(np.sqrt(mean_squared_error(y_test_true, y_test_pred)))
r2 = r2_score(y_test_true, y_test_pred)
print(f"Test MAE (°C): {mae:.6f}")
print(f"Test RMSE (°C): {rmse:.6f}")
print(f"Test R²: {r2:.6f}")

fig, axes = plt.subplots(1, 2, figsize=(11, 4))
ax0, ax1 = axes
ax0.scatter(y_test_true, y_test_pred, alpha=0.35, s=8, c="C0")
lims = [min(y_test_true.min(), y_test_pred.min()), max(y_test_true.max(), y_test_pred.max())]
ax0.plot(lims, lims, "r--", lw=1.5, label="y = x")
ax0.set_xlabel("True T (°C)")
ax0.set_ylabel("Predicted T (°C)")
ax0.set_title("Test: predicted vs. actual")
ax0.legend()
ax0.grid(True, alpha=0.3)
ax0.set_aspect("equal", adjustable="box")

residual_tp = y_test_true - y_test_pred
ax1.hist(residual_tp, bins=40, color="C1", edgecolor="white", alpha=0.9)
ax1.axvline(0.0, color="k", linestyle="--")
ax1.set_xlabel(r"$T_{true} - T_{pred}$ (°C)")
ax1.set_ylabel("Count")
ax1.set_title("Test residual distribution")
ax1.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

fig = plt.figure(figsize=(8, 6))
ax3d = fig.add_subplot(111, projection="3d")
sc3d = ax3d.scatter(
    X_test[:, 0],
    X_test[:, 1],
    X_test[:, 2],
    c=residual_tp,
    cmap="coolwarm",
    s=12,
    alpha=0.55,
)
ax3d.set_xlabel("x")
ax3d.set_ylabel("y")
ax3d.set_zlabel("z")
ax3d.set_title("Test: 3-D sample locations; color = $T_{true} - T_{pred}$ (°C)")
plt.colorbar(sc3d, ax=ax3d, shrink=0.65, label="°C")
plt.tight_layout()
plt.show()

ih = int(np.argmax(y_test_true))
print("Hotspot on test set (largest true T):")
print("  (x, y, z) =", tuple(float(v) for v in X_test[ih, :3]))
print("  (loss_w, tamb_c) =", tuple(float(v) for v in X_test[ih, 3:5]))
print("  T_true  =", float(y_test_true[ih]), "°C")
print("  T_pred  =", float(y_test_pred[ih]), "°C")

# Tamb-colored scatter (uses original test features before scaling)
tamb_test = X_test[:, feature_cols.index("tamb_c")]
fig, ax = plt.subplots(figsize=(6, 5))
sc = ax.scatter(
    y_test_true,
    y_test_pred,
    c=tamb_test,
    cmap="coolwarm",
    alpha=0.45,
    s=12,
    label="",
)
ax.plot(lims, lims, "k--", lw=1)
ax.set_xlabel("True T (°C)")
ax.set_ylabel("Predicted T (°C)")
ax.set_title("Test scatter colored by Tamb (°C)")
plt.colorbar(sc, ax=ax, label="Tamb (°C)")
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
